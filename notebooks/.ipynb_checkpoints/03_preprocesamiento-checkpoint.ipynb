## Dependencias

```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
import joblib
```

## Parámetros

```python
SEED = 42
```

## Lectura de archivos

```python
df=pd.read_excel(r"C:\Users\Morena\TP2\data\raw\smoking_prediction.xlsx") # cargo el dataset
```

## Creación de conjuntos

```python
X = df.drop(columns=["ID","smoking","oral"]) 
y = df["smoking"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=SEED, stratify=y)
```

```python
y_train.value_counts(normalize=True)*100
```

```python
y_test.value_counts(normalize=True)*100
```

```python
X_train.info()
```

## Codificación OHE

```python
ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False) #Inicializo el objeto ohe
```

```python
cat_cols = X_train.select_dtypes(include=["str"]).columns.tolist() ## Categorizo las columnas nominales de tipo STR.
```

```python
cat_cols ## Verifico como quedaron categorizadas
```

```python
%%time
ohe.fit(X_train[cat_cols]) 
## filtro solo por las variables nominales que generamos en la variable "cat_cols" para no codificar todas las variables del DF y lo entreno.
```

```python
X_train_ohe = ohe.transform(X_train[cat_cols]) ## Luego de entrenar, guardo los objetos transformados para ambos conjuntos, Train y Test
X_test_ohe = ohe.transform(X_test[cat_cols])
```

```python
X_train_ohe # obtengo los vectores para cada registro
```

```python
X_test_ohe
```

```python
ohe_columns = ohe.get_feature_names_out(cat_cols) ## Obtengo el nombre de las columnas
```

```python
ohe_columns
```

```python
## Genero un DataFrame con estos datos transformados para el conjunto de entrenamiento, manteniendo el índice (index)
```

```python
X_train_ohe_df = pd.DataFrame(X_train_ohe, columns=ohe_columns, index=X_train.index) 
```

```python
## Genero un DataFrame con estos datos transformados para el conjunto de test, manteniendo el índice (index)
```

```python
X_test_ohe_df = pd.DataFrame(X_test_ohe, columns=ohe_columns, index=X_test.index)
```

```python
X_train_ohe_df
```

```python
X_test_ohe_df
```

```python
## Al conjunto de entrenamiento original X_train le quito las columnas que categorizamos y convertimos a números con
## OHE. Al nuevo DataFrame lo llamaré X_train_num
```

```python
cat_cols
```

```python
X_train_num = X_train.drop(columns=cat_cols)
```

```python
X_train_num
```

```python
## Hago lo mismo con el conjunto de test original (X_train). Al nuevo DataFrame lo llamaré X_test_num
```

```python
X_test_num = X_test.drop(columns=cat_cols)
```

```python
X_test_num
```

```python
## Ahora genero la unión de estos DataFrames, con los que habiamos generado con OHE, para volver a contar con las 
## variables, pero ahora convertidas.
```

```python
X_train_final = pd.concat([X_train_num, X_train_ohe_df], axis=1)
X_test_final = pd.concat([X_test_num, X_test_ohe_df], axis=1)
```

```python
X_train_final
```

```python
X_train_final.info()
```
