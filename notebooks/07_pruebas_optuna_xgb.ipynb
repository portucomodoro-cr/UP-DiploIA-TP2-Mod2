## Dependencias

```python
import joblib
import optuna
import pandas as pd 
from xgboost import XGBClassifier
from sklearn.model_selection import cross_val_score
from sklearn.metrics import classification_report
```

## Parámetros

```python
SEED = 42
```

## Funciones

```python
def objetivo_xgb(trial):
    # Definir el espacio de búsqueda de los hiperparámetros
    n_estimators = trial.suggest_int('n_estimators', 100, 1000)
    max_depth = trial.suggest_int('max_depth', 3, 12)
    learning_rate = trial.suggest_float('learning_rate', 0.001, 0.3, log=True)
    subsample = trial.suggest_float('subsample', 0.5, 1.0)
    colsample_bytree = trial.suggest_float('colsample_bytree', 0.5, 1.0)
    gamma = trial.suggest_float('gamma', 0, 10)
    min_child_weight = trial.suggest_int('min_child_weight', 1, 20)
    reg_alpha = trial.suggest_float('reg_alpha', 1e-5, 10.0, log=True)
    reg_lambda = trial.suggest_float('reg_lambda', 1e-5, 10.0, log=True)
            
    # Crear el modelo con los hiperparámetros sugeridos
    modelo = XGBClassifier(
        n_estimators = n_estimators,
        max_depth = max_depth,
        learning_rate = learning_rate,
        subsample = subsample,
        colsample_bytree = colsample_bytree,
        gamma = gamma,
        min_child_weight = min_child_weight,
        reg_alpha = reg_alpha,
        reg_lambda = reg_lambda,
        random_state = SEED,
        n_jobs=-1
    )
    # Evaluar el modelo (ejemplo: validación cruzada)
    score = cross_val_score(modelo, X_train_final, y_train, n_jobs=-1, cv=3).mean()
    
    return score  # Queremos maximizar la precisión
```

## Lectura de conjuntos de datos procesados

```python
X_train_final = pd.read_excel(r"C:\Users\Morena\TP2\data\processed\X_train_final.xlsx", index_col = 0)
```

```python
y_train = pd.read_excel(r"C:\Users\Morena\TP2\data\processed\y_train.xlsx", index_col = 0).squeeze("columns")
```

```python
X_test_final = pd.read_excel(r"C:\Users\Morena\TP2\data\processed\X_test_final.xlsx", index_col = 0)
```

```python
y_test = pd.read_excel(r"C:\Users\Morena\TP2\data\processed\y_test.xlsx", index_col = 0).squeeze("columns")
```

## Creación de estudios

### Estudio 1

```python
estudio_xgb = optuna.create_study(direction="maximize")
```

```python
estudio_xgb.optimize(objetivo_xgb, n_trials=30)
```

```python
# Ver los mejores resultados
print("Mejores hiperparámetros:", estudio_xgb.best_params)
print("Mejor valor de métrica:", estudio_xgb.best_value)
```

### Estudio 2

```python
estudio_xgb_2 = optuna.create_study(direction="maximize")
```

```python
estudio_xgb_2.optimize(objetivo_xgb, n_trials=100, show_progress_bar=True)
```

```python
print("Mejores hiperparámetros:", estudio_xgb_2.best_params)
print("Mejor valor de métrica:", estudio_xgb_2.best_value)
```
